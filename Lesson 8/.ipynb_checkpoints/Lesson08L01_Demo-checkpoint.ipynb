# Module 8 — Lesson 1: Prompting in Practice (Demo Notebook)
_This notebook illustrates each theme from the lesson in teaching order. It uses a small mock LLM so we can demo concepts without external APIs._

```python

# --- Utilities: a tiny mock LLM for demo purposes ---
import random, re, math
from collections import Counter
import json

def _rng_from_params(temperature=0.7, top_p=0.95, top_k=50, seed=123):
    # derive a seed to make "temperature" etc. visibly affect variety
    s = int((temperature*1000) + (top_p*100) + top_k + seed)
    rng = random.Random(s)
    return rng

import json

def mock_llm(prompt, temperature=0.7, top_p=0.95, top_k=50, shots=None, structure=None):
    """Return a pseudo-response showing effects of params & examples.
    - shots: list of (input, output) pairs for one/few-shot
    - structure: 'bullets', 'json', 'table'
    """
    rng = _rng_from_params(temperature, top_p, top_k)
    vocab = [
        "clear", "concise", "explain", "transformer", "attention", "sequence", "token",
        "compare", "extract", "rewrite", "summarize", "risk", "action", "step",
        "stable", "creative", "nuance", "format", "json", "table", "bullet", "style",
        "audience", "non-technical", "analogy", "structure", "bounds", "null"
    ]
    def sample_word():
        k = max(5, min(int(temperature*len(vocab)) + 5, len(vocab)))
        return rng.choice(vocab[:k])

    example_bias_json = any(isinstance(o, str) and o.strip().startswith("{") for _, o in (shots or []))
    example_bias_bullets = any(isinstance(o, str) and o.strip().startswith("- ") for _, o in (shots or []))

    if structure == "json" or example_bias_json:
        bullets = [f"{sample_word()} {sample_word()} explanation".replace("  "," ").strip() for _ in range(5)]
        resp = {"audience": "non_technical", "bullets": bullets, "analogy": "like a library index"}
        return json.dumps(resp, ensure_ascii=False, indent=2)

    if structure == "bullets" or example_bias_bullets:
        lines = [f"- {sample_word().capitalize()} {sample_word()} idea" for _ in range(5)]
        return "\n".join(lines)

    if structure == "table":
        rows = [["Step","Input","Output"],
                ["1","text","tokens"],
                ["2","tokens","attention"],
                ["3","attn","summary"]]
        return "\n".join([" | ".join(r) for r in rows])

    return " ".join(sample_word() for _ in range(40)).replace("  ", " ").strip()
```

## 1) What is a prompt?
**Goal:** Show how instruction + essential context + explicit output format improves predictability.

```python

vague = "Explain transformers."
clear = (
    "Explain the idea of transformers **to a non-technical audience** in **5 bullets**. "
    "Avoid math. End with **one analogy**. "
    "Return **valid JSON only** with keys: audience (string), bullets (array of 5 strings), analogy (string)."
)

print("Vague prompt:\n", vague)
print("\nClear prompt:\n", clear)
print("\nMock response to vague:")
print(mock_llm(vague))
print("\nMock response to clear (JSON structure):")
print(mock_llm(clear, structure="json"))
```

## 2) Zero / One / Few-shot
Start zero-shot; add one example for structure; add a couple for nuance/style.

```python

task = "Extract product name and price into JSON with keys name (string) and price (number|null). Return JSON only."
one_shot = [("iPhone 15 costs $899", '{"name":"iPhone 15","price":899}')]
few_shot = [
    ("iPhone 15 costs $899", '{"name":"iPhone 15","price":899}'),
    ("Tea set for £25", '{"name":"Tea set","price":25}')
]

print("Zero-shot:")
print(mock_llm(task, structure="json"))
print("\nOne-shot (locks structure):")
print(mock_llm(task, shots=one_shot, structure="json"))
print("\nFew-shot (adds nuance, e.g., currency variants):")
print(mock_llm(task, shots=few_shot, structure="json"))
```

## 3) Show the structure you want
Declare bullets, tables, or JSON; models imitate patterns.

```python

prompt_bullets = "Summarize for non-technical audience in 5 bullets."
print("Bullets:")
print(mock_llm(prompt_bullets, structure="bullets"))

prompt_table = "Explain inference steps in a table with columns Step | Input | Output."
print("\nTable:")
print(mock_llm(prompt_table, structure="table"))
```

## 4) Parameters that matter: temperature, top-p, top-k
We illustrate stability vs creativity with a simple diversity metric.

```python
# Safety guard: define diversity_metric if it's missing
try:
    diversity_metric
except NameError:
    import re
    def diversity_metric(text: str) -> float:
        words = [w for w in re.findall(r"[a-zA-Z]+", text.lower()) if len(w) > 2]
        if not words:
            return 0.0
        return round(len(set(words)) / len(words), 3)
```

```python

# Compare different parameter settings on the same prompt
prompt = "Write a short paragraph about attention in transformers for a non-technical audience."
configs = [
    ("Stable", dict(temperature=0.2, top_p=0.8, top_k=40)),
    ("Balanced", dict(temperature=0.5, top_p=0.9, top_k=60)),
    ("Creative", dict(temperature=0.9, top_p=0.95, top_k=80)),
]

for label, cfg in configs:
    out = mock_llm(prompt, **cfg)
    print(f"\n== {label} - params: {cfg}")
    print(out[:200] + ("..." if len(out)>200 else ""))
    print("diversity:", diversity_metric(out))
```

```python

# Optional: quick plot of diversity (requires matplotlib)
import matplotlib.pyplot as plt

labels = []
divs = []
for label, cfg in [
    ("Stable", dict(temperature=0.2, top_p=0.8, top_k=40)),
    ("Balanced", dict(temperature=0.5, top_p=0.9, top_k=60)),
    ("Creative", dict(temperature=0.9, top_p=0.95, top_k=80)),
]:
    out = mock_llm("Short paragraph on attention.", **cfg)
    labels.append(label)
    divs.append(diversity_metric(out))

plt.figure()
plt.bar(labels, divs)
plt.title("Output Diversity vs. Parameter Presets")
plt.ylabel("Unique-word ratio")
plt.xlabel("Preset")
plt.show()
```

## 5) Compact reasoning: just enough steps
Request 2–4 short steps **only** for multi-part tasks; then give a single labeled answer.

```python

claim = "Claim: The article states the model uses attention."
paragraph = "Paragraph: The method describes weighting tokens by relevance; this weighting is called attention."

no_steps = f"Determine if the claim is supported by the paragraph. Return Supported or Not supported."
with_steps = (
    "Determine if the claim is supported by the paragraph. "
    "Show brief steps (max 3 short lines), then final answer: Supported or Not supported."
)

print("\nNo steps:")
print(mock_llm(no_steps))

print("\nCompact steps + final label:")
print("- Steps:")
print("\n".join([f"{i+1}) step ..." for i in range(3)]))
print("Answer: Supported")
```

## 6) Meet ReAct (concept only)
Reason → Act → Observe → repeat. We simulate a mini trace without real tools.

```python

def mock_search(query):
    # pretend we fetched a definition
    return "Transformers use attention to focus on important parts of the input."

trace = []
trace.append("Thought: Need a plain definition suitable for a non-technical audience.")
trace.append("Action: Search('transformer definition non-technical')")
obs = mock_search("transformer definition non-technical")
trace.append(f"Observation: {obs}")
trace.append("Thought: Summarize in 3 bullets for a non-technical reader.")
answer = "- They focus on the most important words in a sentence.\n- This helps understand meaning better.\n- It works well for long texts."
trace.append("Answer:\n" + answer)

print("\n".join(trace))
```

## 7) Keep context lean
Trim to essentials with a quick checklist: Relevance, Sufficiency, Brevity, Single task.

```python

long_context = """
Please read the following long backstory about our company history from 1998... [200 words of fluff].
Task: Summarize the attached article for non-technical executives in 5 bullets.
Also add a SWOT, translate to French, and include a 500-word memo.
"""

trimmed = """
Task: Summarize the article for non-technical executives in 5 bullets: Context, Key points, Risks, Actions, Open questions.
Bounds: Max 120 words total. No preamble.
"""

print("Before (too much / mixed asks):\n", long_context[:200]+"...")
print("\nAfter (lean, single task):\n", trimmed)
```

## 8) Common pitfalls (and quick fixes)
Mixed tasks, vague audience, no bounds/structure, long examples, missing null policy.

```python

bad = "Summarize and translate and classify the article. Explain transformers."
fixed = (
    "Summarize the article for non-technical readers in 5 bullets (max 120 words).\n"
    "Return valid JSON only: {title, bullets[], language:'en'}. If a field is unknown, use null."
)

print("Before:")
print(bad)
print("\nAfter:")
print(fixed)
```
